# Downstream Torch and PyG export

Artifact v2 directories are Eris canonical storage. Torch and PyG are optional downstream conversions; this notebook does not create repository output or require optional dependencies during its core-only steps.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

import eris

repository = Path.cwd().resolve()
if not (repository / "tests" / "data").is_dir():
    repository = repository.parent
structure = repository / "tests" / "data" / "1AQ2.pdb"
workspace = TemporaryDirectory()
artifact_dir = Path(workspace.name) / "atom_knn_graph"
artifact = eris.encode(
    structure, representation="atom_knn_graph", output=artifact_dir, use_cache=False
)
artifact.validate(strict=True)
```

After installing an appropriate optional Torch stack, use `eris.to_torch(artifact)` for graphs, clouds, matrices, grids, and learned embeddings. To save a downstream file, use `eris export to-torch --artifact-dir artifact_dir --out artifact.pt`.

After installing `eris[pyg]`, `eris.to_pyg(artifact)` and `eris export to-pyg --artifact-dir artifact_dir --out artifact_pyg.pt` are valid only for Artifacts that already expose public graph topology, such as `atom_knn_graph`, `residue_knn_graph`, and `residue_contact_graph`. Clouds, matrices, grids, and learned embeddings are not converted into artificial PyG graphs.
