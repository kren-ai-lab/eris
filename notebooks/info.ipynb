# Artifact inspection

This example creates a disposable Eris artifact and inspects its public metadata.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

import eris

repository = Path.cwd().resolve()
if not (repository / "tests" / "data").is_dir():
    repository = repository.parent
structure = repository / "tests" / "data" / "1AQ2.pdb"
workspace = TemporaryDirectory()
artifact_dir = Path(workspace.name) / "residue_contact_map"
artifact = eris.encode(
    structure, representation="residue_contact_map", output=artifact_dir, use_cache=False
)
artifact.validate(strict=True)
artifact.meta
```
