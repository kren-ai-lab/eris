# Cache inspection

Use the public CLI to inspect cache locations. Clearing cache content is explicit and requires a target plus `--yes`; it is not run by this notebook.

```python
import subprocess
import sys

subprocess.run([sys.executable, "-m", "eris", "cache", "info"], check=True)
subprocess.run([sys.executable, "-m", "eris", "cache", "clear", "--help"], check=True)
```

For a deliberate cleanup, use `eris cache clear --target artifacts --yes`, `--target checkpoints`, or `--target all`. Checkpoint downloads are never automatic.
